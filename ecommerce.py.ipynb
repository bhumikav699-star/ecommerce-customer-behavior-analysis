```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
text_cols = ['Gender', 'City', 'Product_Category', 'Payment_Method', 'Device_Type']
for col in text_cols:
    df[col] = df[col].str.strip().str.title()

df['City'].unique()
df['Product_Category'].unique()
```

```python
df[df['Age'] < 10]                    # unrealistic ages
df[df['Unit_Price'] <= 0]             # zero/negative prices
df[df['Quantity'] <= 0]
df[df['Delivery_Time_Days'] > 30]     # extreme delivery delays
df[df['Customer_Rating'].between(1,5) == False]
```

```python
df['expected_total'] = df['Unit_Price'] * df['Quantity'] - df['Discount_Amount']
mismatch = df[abs(df['expected_total'] - df['Total_Amount']) > 0.01]
mismatch.shape
```

```python
df['Order_Month'] = df['Date'].dt.to_period('M').astype(str)
df['Age_Group'] = pd.cut(df['Age'], bins=[17,25,35,45,55,100],
                          labels=['18-25','26-35','36-45','46-55','56+'])
```

```python
df.drop(columns=['expected_total'], inplace=True)  # drop helper column
df.to_csv('ecommerce_cleaned.csv', index=False)
```
